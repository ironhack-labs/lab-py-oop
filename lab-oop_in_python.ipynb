### Exercise 4: Class Inheritance with Method Overriding
Create a `Bus` class that inherits from the `Vehicle` class. Override the `fare()` method to include an extra charge for maintenance.

```python
class Vehicle:
    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage

    def fare(self):
        # Basic fare calculation: assume rubles 1.5 per km for example
        return self.mileage * 1.5


class Bus(Vehicle):
    def fare(self):
        # Base fare from Vehicle
        base_fare = super().fare()
        # Add 10% maintenance charge
        maintenance_charge = base_fare * 0.10
        return base_fare + maintenance_charge


# Example usage
if __name__ == "__main__":
    school_bus = Bus(80, 1200)
    print("Total Bus Fare (including maintenance): Rubles.", school_bus.fare())
```

### Exercise 6: Class Inheritance with Default Fare Calculation
Create a `Bus` child class that inherits from the `Vehicle` class. The default fare charge of any vehicle is `seating capacity * 100`. If the vehicle is a bus instance, add an extra 10% on the full fare as a maintenance charge.

```python
class Vehicle:
    def __init__(self, max_speed, mileage, capacity):
        self.max_speed = max_speed
        self.mileage = mileage
        self.capacity = capacity

    def fare(self):
        return self.capacity * 100


class Bus(Vehicle):
    def fare(self):
        base_fare = super().fare()
        maintenance_charge = base_fare * 0.10
        return base_fare + maintenance_charge


# Example usage
if __name__ == "__main__":
    school_bus = Bus(80, 1200, 50)
    print("Total Bus Fare (with maintenance): RMB.", school_bus.fare())
```

### Exercise 7: Check Type of an Object
Write a program to determine which class a given object belongs to.

```python
class Vehicle:
    pass

class Bus(Vehicle):
    pass

class Car(Vehicle):
    pass

# Example objects
bus1 = Bus()
car1 = Car()
vehicle1 = Vehicle()

# Function to print the class of an object
def print_class(obj):
    print(f"The object belongs to the class: {type(obj).__name__}")

# Test the function
print_class(bus1)     # Output: Bus
print_class(car1)     # Output: Car
print_class(vehicle1) # Output: Vehicle
```

### Exercise 8: Check Instance of Class 
Determine if `school_bus` is also an instance of the `Vehicle` class.


```python
class Vehicle:
    pass

class Bus(Vehicle):
    pass

# Create an instance of Bus
school_bus = Bus()

# Check if school_bus is also an instance of Vehicle
if isinstance(school_bus, Vehicle):
    print("school_bus is an instance of the Vehicle class.")
else:
    print("school_bus is NOT an instance of the Vehicle class.")
```
