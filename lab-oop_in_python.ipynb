### Exercise 1: Create a Class with Instance Attributes
Write a Python program to create a `Vehicle` class with `max_speed` and `mileage` instance attributes.

```python
class Vehicle:
    def __init__(self, max_speed, mileage): # instantiate the class with __init__ method
        self.max_speed = max_speed # instance attributes called with the self.
        self.mileage = mileage
    pass

modelX = Vehicle(200, 10) # instance the class "modelX"

print(modelX.max_speed, modelX.mileage)
```

### Exercise 4: Class Inheritance with Method Overriding
Create a `Bus` class that inherits from the `Vehicle` class. Override the `fare()` method to include an extra charge for maintenance.

```python
class Vehicle:
    def fare(self):
        return "Base fare"

class Bus(Vehicle):
    def fare(self): # <-- to override the fare method I call the function inside the child class.
        return "Base fare with extra charge" # <- this is now "overriding" the result.

school_bus = Bus() # call child bus class and assign to school_bus variable to allow us to print child result with overiding function "fare"
print(school_bus.fare())  # Expected output: "Base fare with extra charge"
```

### Exercise 5: Define a Class Attribute
Define a property that must have the same value for every class instance (object). Set a default value for `color`.

```python

class Vehicle:
    color = "White" # <-- The color is defined here on the class not on "self" making it the default value assign to the class.

    def __init__(self, name, max_speed, mileage):
        self.name = name
        self.max_speed = max_speed
        self.mileage = mileage

school_bus = Vehicle("School Volvo", 180, 12)
print(school_bus.color)  # Expected output: "White"
```

### Exercise 6: Class Inheritance with Default Fare Calculation
Create a `Bus` child class that inherits from the `Vehicle` class. The default fare charge of any vehicle is `seating capacity * 100`. If the vehicle is a bus instance, add an extra 10% on the full fare as a maintenance charge.

```python
class Vehicle:
    def __init__(self, name, mileage, capacity):
        self.name = name
        self.mileage = mileage
        self.capacity = capacity

    def fare(self):
         return self.capacity * 100



class Bus(Vehicle):
    def fare(self):
        base_fare = super().fare() # To add the fare I needed to make a base_fare variable to call the fare function from the parent class  using the super() function.
        maintenance_charge = base_fare * 0.10 # Then create a maintenance variable and add the base_fare variable to create the charge.
        return base_fare + maintenance_charge # we now add the charge to the base fare which is then added to the final result to include the charge.

school_bus = Bus("School Volvo", 12, 50)
print("Total Bus fare is:", school_bus.fare())  # Expected output: Total Bus fare is: (calculated amount)
```
