### Exercise 4: Class Inheritance with Method Overriding
Create a `Bus` class that inherits from the `Vehicle` class. Override the `fare()` method to include an extra charge for maintenance.

```python
# Define parent class Vehicle with a fare() method
class Vehicle:
    
    def fare(self):
        return "Base fare"

class Bus(Vehicle): #Bus inherits from Vehicle, but needs to override fare() method
    # Hint: Override fare method to include extra charges
    def fare(self):
        base_fare = super().fare()  #calls the parent Vehicle version of fare()
        new_fare = base_fare + " with extra charge" # adds/overrides parent's method
        return new_fare

school_bus = Bus()
print(school_bus.fare())  # Expected output: "Base fare with extra charge"
```

### Exercise 6: Class Inheritance with Default Fare Calculation
Create a `Bus` child class that inherits from the `Vehicle` class. The default fare charge of any vehicle is `seating capacity * 100`. If the vehicle is a bus instance, add an extra 10% on the full fare as a maintenance charge.

```python
# combine inheritance, method overriding, and a default fare formula
class Vehicle:              # Base (parent) class for all types of vehicles
    
    def __init__(self, name, mileage, capacity):     # has instance variables: name, mileage and capacity
        self.name = name                              
        self.mileage = mileage
        self.capacity = capacity

    def fare(self): #default fare calculation       # the method fare() returns capacity *100 - default fare formula
        return self.capacity * 100
    
# Child class (Bus)         
class Bus(Vehicle):             # Bus inherits from Vehicle
    # Hint: Override fare method to include maintenance charge
    def fare(self):
        base_fare = super().fare() #get base from parent class
        new_fare = base_fare + (0.1 * base_fare) # new fare wirh addirional maintainance charge
        return new_fare #return overriden new_fare


# Create a Bus object
school_bus = Bus("School Volvo", 12, 50)
# print the total fare
print(f"Total Bus fare is: {school_bus.fare()}")  

# line has two parts 
    #1. school_bus.fare() - calls a method named fare(), that belongs to school_bus object in the Bus class; calculates; returns 
    #2. f before the string (firmatted string literal f-string); anything inside {...} is evaluated first, then inserted into the str.
# Expected output: Total Bus fare is: (calculated amount)
```
