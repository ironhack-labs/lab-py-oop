### Exercise 4: Class Inheritance with Method Overriding
Create a `Bus` class that inherits from the `Vehicle` class. Override the `fare()` method to include an extra charge for maintenance.

```python
# Define the Vehicle class with a base fare method
class Vehicle:
    def fare(self):
        return "Base fare"  # Base fare for any vehicle

# Define the Bus class that inherits from Vehicle and adds extra charge
class Bus(Vehicle):
    def fare(self):
        return super().fare() + " with extra charge"  # Add maintenance charge

# Create an instance of Bus
school_bus = Bus()

# Print the fare for the Bus instance
print(school_bus.fare())  # Expected output: "Base fare with extra charge"

```

### Exercise 6: Class Inheritance with Default Fare Calculation
Create a `Bus` child class that inherits from the `Vehicle` class. The default fare charge of any vehicle is `seating capacity * 100`. If the vehicle is a bus instance, add an extra 10% on the full fare as a maintenance charge.

```python
class Vehicle:
    def __init__(self, name, mileage, capacity):
        self.name = name
        self.mileage = mileage
        self.capacity = capacity

    def fare(self):
        return self.capacity * 100

class Bus(Vehicle):
    def fare(self):
        base_fare = super().fare()
        return base_fare + (base_fare * 0.10)

school_bus = Bus("School Volvo", 12, 50)
print("Total Bus fare is:", school_bus.fare())

```

### Exercise 7: Check Type of an Object
Write a program to determine which class a given object belongs to.

```python
# Define the Vehicle class with constructor and fare method
class Vehicle:
    def __init__(self, name, mileage, capacity):
        self.name = name        # Name of the vehicle
        self.mileage = mileage  # Mileage of the vehicle
        self.capacity = capacity  # Seating capacity

    def fare(self):
        return self.capacity * 100  # Base fare calculation

# Define the Bus class that inherits from Vehicle and adds maintenance charge
class Bus(Vehicle):
    def fare(self):
        base_fare = super().fare()       # Call the base fare from Vehicle
        return base_fare + (base_fare * 0.10)  # Add 10% maintenance charge

# Create an instance of Bus
school_bus = Bus("School Volvo", 12, 50)

# Print type and fare for verification
print(type(school_bus))           # Expected output: <class '__main__.Bus'>
print("Total Bus fare is:", school_bus.fare())  # Expected output: 5500.0

```
